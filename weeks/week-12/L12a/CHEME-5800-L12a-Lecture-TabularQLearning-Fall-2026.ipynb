## Update rule

After observing $(s_t,a_t,r_{t+1},s_{t+1})$, tabular Q-learning applies

$$Q(s_t,a_t)\leftarrow Q(s_t,a_t)+\alpha\left[r_{t+1}+\gamma\max_a Q(s_{t+1},a)-Q(s_t,a_t)\right].$$

The bracketed quantity is the temporal-difference error. At a terminal next state, the future-value term is zero.


## Exploration contract

An $\epsilon$-greedy behavior policy chooses a random action with probability $\epsilon$ and otherwise chooses a current maximizer. A decaying schedule can shift from broad exploration toward exploitation, but premature decay leaves state-action pairs poorly estimated.

## Summary

Q-learning is model-free in the specific sense that its update does not require the transition matrix. It still requires state/action definitions, rewards, terminal semantics, sampled interaction, and enough coverage to learn useful values.
