## Practicum preparation workflow

1. Restate the engineering question and define success.
2. Make inputs, units, dimensions, and identifiers explicit.
3. Choose a method whose assumptions match the problem.
4. Build the smallest reference case before the full calculation.
5. Test invariants and at least one failure path.
6. Record data provenance and the exact environment.
7. Interpret the result and state limitations.

The practicum problem itself is released through the separate assessment repository. This notebook supplies the integration method, not the answer-bearing assessment artifact.
