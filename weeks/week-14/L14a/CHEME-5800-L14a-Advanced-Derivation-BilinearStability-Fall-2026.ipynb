# Derivation: Stability of the Bilinear Discretization
In this companion derivation, we will show that the bilinear discretization keeps the stability of a continuous linear state space model, for any time step. We'll find the eigenvalues of the discrete state matrix, compute their magnitudes, and compare them with one. The [main lecture](CHEME-5800-L14a-Lecture-LinearStateSpaceModels-Fall-2026.ipynb) states this result and uses the recursion to solve the forced model.

> __Learning Objectives:__
>
> By the end of this derivation, you should be able to:
>
> * __Relate the eigenvalues of the two state matrices:__ The discrete state matrix is built from the continuous state matrix, and it has the same eigenvectors. We'll derive the formula that maps each eigenvalue of the continuous matrix to an eigenvalue of the discrete matrix.
> * __Write the recursion as a sum of modes:__ Without an input, the recursion multiplies the state by the discrete state matrix at every step. We'll expand the initial state in the eigenvectors and show that each step multiplies every mode by a constant factor.
> * __Compare the stability of the model and the recursion:__ A mode of the recursion decays when the magnitude of its eigenvalue is less than one. We'll show that this happens exactly when the matching eigenvalue of the continuous model has a negative real part, for any time step.

Let's get started!

___

## The Discrete State Matrix
The bilinear discretization replaces the continuous model $\dot{\mathbf{x}}=\mathbf{A}\mathbf{x}+\mathbf{B}\mathbf{u}$ with a recursion on a grid of times $t_{k}=k\Delta t$, with a step $\Delta t>0$:

$$
\mathbf{x}_{k} = \bar{\mathbf{A}}\mathbf{x}_{k-1}+\bar{\mathbf{B}}\mathbf{u}_{k},\qquad
\bar{\mathbf{A}} = \left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)^{-1}\left(\mathbf{I}+\tfrac{\Delta t}{2}\mathbf{A}\right),\qquad
\bar{\mathbf{B}} = \left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)^{-1}\Delta t\,\mathbf{B}.
$$

The inverse of $\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}$ exists when no eigenvalue of $\mathbf{A}$ equals $2/\Delta t$. As in the lecture, we assume that $\mathbf{A}$ has $n$ linearly independent eigenvectors $\mathbf{v}_{i}$, with $\mathbf{A}\mathbf{v}_{i}=\lambda_{i}\mathbf{v}_{i}$. We determine the stability of the recursion from the eigenvalues of $\bar{\mathbf{A}}$.

___

## Eigenvalues of the Discrete State Matrix
Take one eigenvector $\mathbf{v}$ of $\mathbf{A}$, with $\mathbf{A}\mathbf{v}=\lambda\mathbf{v}$, and let $z=\Delta t\,\lambda/2$. First, we find how each factor of $\bar{\mathbf{A}}$ acts on $\mathbf{v}$:

$$
\begin{aligned}
\left(\mathbf{I}+\tfrac{\Delta t}{2}\mathbf{A}\right)\mathbf{v} &= \mathbf{v}+\tfrac{\Delta t}{2}\mathbf{A}\mathbf{v} &&\quad\text{(distribute, using }\mathbf{I}\mathbf{v}=\mathbf{v}\text{)}\\
&= \mathbf{v}+\tfrac{\Delta t}{2}\lambda\,\mathbf{v} &&\quad\text{(use }\mathbf{A}\mathbf{v}=\lambda\mathbf{v}\text{)}\\
&= (1+z)\,\mathbf{v} &&\quad\text{(factor out }\mathbf{v}\text{, with }z=\Delta t\,\lambda/2\text{)}\\
\left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)\mathbf{v} &= (1-z)\,\mathbf{v} &&\quad\text{(same three steps, with a minus sign)}\\
\mathbf{v} &= (1-z)\left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)^{-1}\mathbf{v} &&\quad\text{(multiply both sides by }\left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)^{-1}\text{)}\\
\left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)^{-1}\mathbf{v} &= \frac{1}{1-z}\,\mathbf{v} &&\quad\text{(divide by }1-z\text{, which is not zero because }\lambda\neq 2/\Delta t\text{)}
\end{aligned}
$$

Then, we apply the two factors to $\mathbf{v}$, one after the other:

$$
\begin{aligned}
\bar{\mathbf{A}}\mathbf{v} &= \left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)^{-1}\left(\mathbf{I}+\tfrac{\Delta t}{2}\mathbf{A}\right)\mathbf{v} &&\quad\text{(definition of }\bar{\mathbf{A}}\text{)}\\
&= \left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)^{-1}(1+z)\,\mathbf{v} &&\quad\text{(first factor)}\\
&= (1+z)\left(\mathbf{I}-\tfrac{\Delta t}{2}\mathbf{A}\right)^{-1}\mathbf{v} &&\quad\text{(move the number }1+z\text{ to the front)}\\
&= \underbrace{\frac{1+z}{1-z}}_{\displaystyle\bar{\lambda}}\,\mathbf{v} &&\quad\text{(second factor)}
\end{aligned}
$$

The vector $\mathbf{v}$ is an eigenvector of both matrices, but its eigenvalue changes from $\lambda$ for $\mathbf{A}$ to $\bar{\lambda}$ for $\bar{\mathbf{A}}$. Each eigenvalue $\lambda_{i}$ of $\mathbf{A}$ therefore gives the eigenvalue $\bar{\lambda}_{i}=(1+z_{i})/(1-z_{i})$ of $\bar{\mathbf{A}}$, with $z_{i}=\Delta t\,\lambda_{i}/2$.

___

## Modes of the Recursion
Without an input, the recursion is $\mathbf{x}_{k}=\bar{\mathbf{A}}\mathbf{x}_{k-1}$, and applying it $k$ times gives $\mathbf{x}_{k}=\bar{\mathbf{A}}^{k}\mathbf{x}_{0}$. As in the lecture, we expand the initial state in the eigenvectors, $\mathbf{x}_{0}=c_{1}\mathbf{v}_{1}+\dots+c_{n}\mathbf{v}_{n}$, with the coefficients $\mathbf{c}=\mathbf{V}^{-1}\mathbf{x}_{0}$, where the columns of $\mathbf{V}$ are the eigenvectors $\mathbf{v}_{i}$:

$$
\begin{aligned}
\mathbf{x}_{k} &= \bar{\mathbf{A}}^{k}\left(c_{1}\mathbf{v}_{1}+\dots+c_{n}\mathbf{v}_{n}\right) &&\quad\text{(expand }\mathbf{x}_{0}\text{ in the eigenvectors)}\\
&= c_{1}\bar{\mathbf{A}}^{k}\mathbf{v}_{1}+\dots+c_{n}\bar{\mathbf{A}}^{k}\mathbf{v}_{n} &&\quad\text{(multiply each term by }\bar{\mathbf{A}}^{k}\text{)}\\
&= c_{1}\bar{\lambda}_{1}^{k}\mathbf{v}_{1}+\dots+c_{n}\bar{\lambda}_{n}^{k}\mathbf{v}_{n} &&\quad\text{(apply }\bar{\mathbf{A}}\mathbf{v}_{i}=\bar{\lambda}_{i}\mathbf{v}_{i}\text{ a total of }k\text{ times)}
\end{aligned}
$$

Each term $c_{i}\bar{\lambda}_{i}^{k}\mathbf{v}_{i}$ is a mode of the recursion, and each step multiplies it by the factor $\bar{\lambda}_{i}$. Because $|\bar{\lambda}_{i}^{k}|=|\bar{\lambda}_{i}|^{k}$, the mode decays to zero when $|\bar{\lambda}_{i}|<1$.

___

## The Magnitude of the Discrete Eigenvalue
To compare $|\bar{\lambda}|$ with one, we write the eigenvalue as $\lambda=\alpha+i\beta$, with real part $\alpha$ and imaginary part $\beta$, as in the lecture. Then $z=\tfrac{\Delta t}{2}\alpha+i\,\tfrac{\Delta t}{2}\beta$. The squared magnitude of a complex number is the square of its real part plus the square of its imaginary part:

$$
\begin{aligned}
|1+z|^{2} &= \left(1+\tfrac{\Delta t}{2}\alpha\right)^{2}+\left(\tfrac{\Delta t}{2}\beta\right)^{2} &&\quad\text{(real part }1+\tfrac{\Delta t}{2}\alpha\text{, imaginary part }\tfrac{\Delta t}{2}\beta\text{)}\\
|1-z|^{2} &= \left(1-\tfrac{\Delta t}{2}\alpha\right)^{2}+\left(\tfrac{\Delta t}{2}\beta\right)^{2} &&\quad\text{(real part }1-\tfrac{\Delta t}{2}\alpha\text{, imaginary part }-\tfrac{\Delta t}{2}\beta\text{)}\\
|1+z|^{2}-|1-z|^{2} &= \left(1+\tfrac{\Delta t}{2}\alpha\right)^{2}-\left(1-\tfrac{\Delta t}{2}\alpha\right)^{2} &&\quad\text{(the }\beta\text{ terms cancel)}\\
&= \left(1+\Delta t\,\alpha+\tfrac{\Delta t^{2}}{4}\alpha^{2}\right)-\left(1-\Delta t\,\alpha+\tfrac{\Delta t^{2}}{4}\alpha^{2}\right) &&\quad\text{(expand both squares)}\\
&= 2\,\Delta t\,\alpha &&\quad\text{(the }1\text{ and }\alpha^{2}\text{ terms cancel)}
\end{aligned}
$$

Because $\bar{\lambda}=(1+z)/(1-z)$, its squared magnitude is $|\bar{\lambda}|^{2}=|1+z|^{2}/|1-z|^{2}$. The mode of the recursion decays when this ratio is less than one:

$$
\begin{aligned}
|\bar{\lambda}|<1 &\iff |1+z|^{2}<|1-z|^{2} &&\quad\text{(square both sides, then multiply by }|1-z|^{2}>0\text{)}\\
&\iff 2\,\Delta t\,\alpha<0 &&\quad\text{(subtract }|1-z|^{2}\text{ and use the result above)}\\
&\iff \alpha<0 &&\quad\text{(divide by }2\,\Delta t>0\text{)}
\end{aligned}
$$

The condition $\alpha<0$ is the same condition that makes the mode decay in the continuous model. The same steps, with $=$ or $>$ in place of $<$, show that $\alpha=0$ gives $|\bar{\lambda}|=1$ and $\alpha>0$ gives $|\bar{\lambda}|>1$. Each mode of the recursion therefore decays, keeps its size, or grows exactly when the matching mode of the continuous model does.

## Summary
We showed that the bilinear discretization keeps the stability of a continuous linear state space model, for any time step.

> __Key Takeaways:__
>
> * __The two state matrices share their eigenvectors:__ We showed that each factor of the discrete state matrix multiplies an eigenvector of the continuous state matrix by a number. The eigenvector stays the same, while its eigenvalue changes to a new value that depends on the time step.
>
> * __The recursion is a sum of modes:__ Without an input, we expanded the initial state in the eigenvectors and applied the recursion one step at a time. Each step multiplies every mode by its discrete eigenvalue, and a mode decays when that eigenvalue has magnitude less than one.
>
> * __Stability does not depend on the time step:__ We showed that a discrete eigenvalue has magnitude less than one exactly when the matching continuous eigenvalue has a negative real part. A stable model therefore gives a stable recursion for any time step.

The time step therefore sets only the accuracy of the recursion, not its stability.
