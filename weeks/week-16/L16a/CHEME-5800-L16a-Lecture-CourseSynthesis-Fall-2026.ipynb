## A complete computational claim


```julia
policy_claim = (
    question = "Which operating policy maximizes expected return?",
    representation = "finite Markov decision process",
    method = "value iteration",
    evidence = "Bellman residual plus independent rollout evaluation",
    limitations = "transition probabilities and rewards are assumed known and stationary",
)
validate_computational_contract(policy_claim)
```

A numerical answer without evidence is incomplete. Evidence without limitations is easy to overinterpret. The computational contract forces the engineering claim, not just the code, to be inspectable.


```julia
incomplete_claim = (
    question = "Will this machine fail?",
    representation = "sensor feature table",
    method = "logistic regression",
    evidence = "",
    limitations = "",
)
validate_computational_contract(incomplete_claim)
```

## Final synthesis prompts

1. Which representation choice most strongly constrained your result this semester?
2. Which invariant or reference case gave you the most confidence?

4. Which system boundary introduced the most consequential failure mode?
