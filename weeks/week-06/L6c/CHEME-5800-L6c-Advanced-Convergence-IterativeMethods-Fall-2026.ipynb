# Advanced: Where does the spectral radius convergence condition come from?
The [general iterative method](CHEME-5800-L6c-Lecture-GeneralIterativeMethod-Fall-2026.ipynb) converges from every initial guess when the spectral radius of its iteration matrix is less than one. Here we derive the error recurrence and explain why the eigenvalues determine whether every initial error eventually disappears. The argument concerns exact arithmetic.

> __Learning Objectives:__
>
> By the end of this notebook, you should be able to:
> - **Derive error propagation:** Relate the error after repeated updates to powers of the iteration matrix.
> - **Explain the convergence criterion:** Establish why all eigenvalues must lie strictly inside the unit circle.
> - **Interpret convergence correctly:** Distinguish eventual convergence from error reduction at every iteration.

___


## From the matrix splitting to the error recurrence
Consider $\mathbf{A}\mathbf{x}=\mathbf{b}$, where $\mathbf{A}\in\mathbb{R}^{n\times n}$ is nonsingular, $\mathbf{x}$ contains the unknowns, and $\mathbf{b}$ is the right-hand side. Choose a splitting $\mathbf{A}=\mathbf{M}-\mathbf{N}$ with $\mathbf{M}$ nonsingular. The stationary iteration is:

$$
\begin{aligned}
\mathbf{x}^{(k+1)}
&=\mathbf{M}^{-1}(\mathbf{b}+\mathbf{N}\mathbf{x}^{(k)})\\
&=\underbrace{\mathbf{M}^{-1}\mathbf{b}}_{\mathbf{c}}
+\underbrace{\mathbf{M}^{-1}\mathbf{N}}_{\mathbf{G}}\mathbf{x}^{(k)}.
\end{aligned}
$$

Here $\mathbf{G}$ is the iteration matrix and $\mathbf{c}$ is constant. The exact solution $\mathbf{x}^{\star}$ is a fixed point because:

$$
(\mathbf{M}-\mathbf{N})\mathbf{x}^{\star}=\mathbf{b}
\quad\Longrightarrow\quad
\mathbf{x}^{\star}=\mathbf{c}+\mathbf{G}\mathbf{x}^{\star}.
$$

This fixed point is unique: $\mathbf{I}-\mathbf{G}=\mathbf{M}^{-1}\mathbf{A}$ is nonsingular, where $\mathbf{I}$ is the identity matrix. Define the solution error by $\mathbf{e}^{(k)}=\mathbf{x}^{(k)}-\mathbf{x}^{\star}$. Subtracting the fixed-point equation gives:

$$
\begin{aligned}
\mathbf{e}^{(k+1)}
&=(\mathbf{c}+\mathbf{G}\mathbf{x}^{(k)})-(\mathbf{c}+\mathbf{G}\mathbf{x}^{\star})\\
&=\mathbf{G}(\mathbf{x}^{(k)}-\mathbf{x}^{\star})
=\mathbf{G}\mathbf{e}^{(k)},\\[3pt]
\mathbf{e}^{(1)}&=\mathbf{G}\mathbf{e}^{(0)},\qquad
\mathbf{e}^{(2)}=\mathbf{G}^{2}\mathbf{e}^{(0)},\qquad
\boxed{\mathbf{e}^{(k)}=\mathbf{G}^{k}\mathbf{e}^{(0)}}.
\end{aligned}
$$

Thus, convergence **from every initial guess** is equivalent to $\mathbf{G}^{k}\to\mathbf{0}$. One direction follows by multiplying by any initial error; the other follows by choosing each coordinate unit vector as the initial error, so every column of $\mathbf{G}^{k}$ must tend to zero.

___


## Why the spectral radius decides convergence
Let $\lambda_i$ denote the possibly complex eigenvalues of $\mathbf{G}$. The [matrix-power convergence theorem](https://nhigham.com/2024/01/12/what-is-the-spectral-radius-of-a-matrix/) states:

> __Spectral-radius criterion:__
>
> For a finite square iteration matrix, the following conditions are equivalent:
>
> $$
> \mathbf{G}^{k}\longrightarrow\mathbf{0}
> \quad\Longleftrightarrow\quad
> \rho(\mathbf{G})=\max_i|\lambda_i|<1.
> $$
>
> Combined with the error recurrence, this characterizes convergence to the exact solution from every initial guess.

__Why the condition is necessary.__ For an eigenvector $\mathbf{v}\ne\mathbf{0}$ with eigenvalue $\lambda$, repeated multiplication gives:

$$
\mathbf{G}^{k}\mathbf{v}=\lambda^{k}\mathbf{v},
\qquad
\|\mathbf{G}^{k}\|_2
\ge\frac{\|\mathbf{G}^{k}\mathbf{v}\|_2}{\|\mathbf{v}\|_2}
=|\lambda|^{k}.
$$

Here the matrix norm is the induced Euclidean norm. If any $|\lambda|\ge1$, this lower bound does not tend to zero, so neither can $\mathbf{G}^{k}$. This argument also covers complex eigenvalues of a real matrix.

__Why the condition is sufficient.__ If $\mathbf{G}$ has a complete eigenvector basis, write $\mathbf{G}=\mathbf{V}\boldsymbol{\Lambda}\mathbf{V}^{-1}$, where the columns of $\mathbf{V}$ are eigenvectors and $\boldsymbol{\Lambda}$ contains their eigenvalues. Then:

$$
\mathbf{G}^{k}
=\mathbf{V}\,\operatorname{diag}(\lambda_1^k,\ldots,\lambda_n^k)\,\mathbf{V}^{-1}
\longrightarrow\mathbf{0}
\qquad\text{when every }|\lambda_i|<1.
$$

A complete eigenvector basis is not required. In the general case, use the Jordan form $\mathbf{G}=\mathbf{V}\mathbf{J}\mathbf{V}^{-1}$ over the complex numbers, with $\mathbf{V}$ invertible. Each size-$p$ Jordan block has the form $\mathbf{B}=\lambda\mathbf{I}+\mathbf{K}$, where $\mathbf{K}$ has ones immediately above the diagonal and zeros elsewhere, so $\mathbf{K}^{p}=\mathbf{0}$. For $\lambda\ne0$ and $k\ge p$, its power is:

$$
\mathbf{B}^{k}
=\sum_{j=0}^{p-1}\binom{k}{j}\lambda^{k-j}\mathbf{K}^{j}.
$$

For each fixed $j$, the binomial coefficient grows at most polynomially in $k$, while $|\lambda|^{k-j}$ decays geometrically when $|\lambda|<1$. Every term therefore tends to zero. If $\lambda=0$, the block power is zero for $k\ge p$. All blocks tend to zero, giving $\mathbf{J}^{k}\to\mathbf{0}$ and hence $\mathbf{G}^{k}\to\mathbf{0}$.

The phrase “every initial guess” matters. For $\mathbf{G}=\operatorname{diag}(1/2,2)$, an initial error with zero second component decays, even though the spectral radius is two. That special initial error does not establish convergence of the method for arbitrary guesses.

## Convergence need not be monotone
Eigenvalues describe eventual decay, but they do not require the error norm to decrease at every step. Consider the following iteration matrix and initial error:

$$
\mathbf{G}=\begin{bmatrix}1/2&2\\0&1/2\end{bmatrix},
\qquad
\mathbf{e}^{(0)}=\begin{bmatrix}0\\1\end{bmatrix},
\qquad
\mathbf{e}^{(1)}=\begin{bmatrix}2\\1/2\end{bmatrix}.
$$

Both eigenvalues are $1/2$, yet the first error norm increases from $1$ to $\sqrt{17}/2\approx2.06$. Repeated multiplication still gives:

$$
\mathbf{G}^{k}=2^{-k}\begin{bmatrix}1&4k\\0&1\end{bmatrix}
\longrightarrow\mathbf{0}.
$$

The factor $k$ can cause temporary growth, but geometric decay eventually dominates it. A bound $\|\mathbf{G}\|_2<1$ would guarantee contraction at every step in the Euclidean norm; it is sufficient, but is not necessary for convergence. The [companion example](CHEME-5800-L6c-Example-FunWithIterativeSolvers-Fall-2026.ipynb) checks the residual $\mathbf{r}^{(k)}=\mathbf{b}-\mathbf{A}\mathbf{x}^{(k)}=-\mathbf{A}\mathbf{e}^{(k)}$, whose relationship to solution error depends on the conditioning of $\mathbf{A}$.

## Summary
We connected the matrix splitting to the decay of the solution error.

> __Key Takeaways:__
>
> - **Matrix powers propagate the error:** We subtracted the fixed-point equation from the update and expressed every later error in terms of the initial error.
> - **Every eigenvalue must decay:** We established the spectral-radius criterion for convergence from every initial guess, including matrices without a complete eigenvector basis.
> - **Eventual convergence allows temporary growth:** We examined an iteration whose error initially increases before decaying, distinguishing convergence from contraction at every step.

These results explain the convergence condition used to assess Jacobi, Gauss–Seidel, and successive over-relaxation in the lecture.
