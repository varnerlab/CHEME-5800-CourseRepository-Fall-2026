# Derivation: Flux Balance Analysis (FBA) Material Balances
In this companion derivation, we will derive the material-balance constraint used in flux balance analysis (FBA). We'll start with an open mole balance, distinguish concentrations per unit volume from amounts per unit biomass, and show which assumptions lead to [the standard formulation of Orth, Thiele, and Palsson](https://pmc.ncbi.nlm.nih.gov/articles/PMC3108565/). The [main lecture](CHEME-5800-L6a-Lecture-FluxBalanceAnalysis-Fall-2026.ipynb) combines this constraint with flux bounds and an objective function.

> __Learning Objectives:__
>
> By the end of this derivation, you should be able to:
>
> * __Derive species concentration balances:__ A species concentration balance is an open mole balance written per unit volume or per unit biomass, and the two bases carry different accumulation terms. We'll derive both forms from the same amount balance and keep the units of the concentrations and reaction rates consistent throughout.
> * __Apply steady-state approximations:__ The growth-dilution term is the production needed to keep a metabolite's amount per unit biomass constant as the cells grow. We'll derive this term, then show which assumptions about balanced growth and dilution reduce the dynamic balance to the conventional flux balance constraint.
> * __Explain exchange reactions:__ Exchange reactions connect the modeled metabolite pools to surroundings that the model does not balance, which lets the stoichiometric equations describe an open system. We'll distinguish transport from exchange and separate the intracellular balance assumptions from the reactor operating conditions.

## Theory: Material Balances
Suppose we have a well-mixed system with physical volume $V>0$, measured in $\mathrm{L}$. Inside this system, we have a reaction set $\mathcal{R}$ and a metabolite set $\mathcal{M}$. A stream set $\mathcal{F}$ connects the system to its surroundings. We will first write the balance on a physical-volume basis, then repeat the calculation on a biomass basis for intracellular metabolites.

> __Material Balances:__
>
> A material balance for species $i\in\mathcal{M}$ accounts for accumulation, net generation, and transport across the system boundary:
>
> $$
> \text{Accumulation} = \text{Generation} + \text{Transport In} - \text{Transport Out}.
> $$
>
> Accumulation is the rate of change of the amount of species $i$ inside the system. Net generation accounts for its production and consumption by chemical reactions. Transport accounts for material entering or leaving the selected system.

Let $n_i$ denote the amount of species $i$ in $\mathrm{mmol}$, $\dot n_{i,s}$ its nonnegative molar flow rate in stream $s$ in $\mathrm{mmol\,h^{-1}}$, and $\dot n_{G,i}$ its net generation rate in the same units. Define the stream-direction coefficient $d_s=+1$ for an inlet and $d_s=-1$ for an outlet. The open species mole balance is given by:

$$
\frac{dn_i}{dt}
= \sum_{s\in\mathcal{F}}d_s\dot n_{i,s}+\dot n_{G,i},
\qquad i\in\mathcal{M}.
$$

### Dynamic species concentration balances
For a bulk fluid stream, let $c_{i,s}$ denote the species concentration in $\mathrm{mmol\,L^{-1}}$ and $\dot V_s\geq0$ the volumetric flow rate in $\mathrm{L\,h^{-1}}$. Inside the well-mixed system, let $c_i=n_i/V$ be the concentration in $\mathrm{mmol\,L^{-1}}$. We use lowercase $c_i$ for this volume basis and reserve uppercase $C_i$ for the biomass basis introduced next.

The volumetric reaction rate $v_j$ has units $\mathrm{mmol\,L^{-1}\,h^{-1}}$, or concentration per time. The dimensionless stoichiometric coefficient $\sigma_{ij}$ is positive when reaction $j$ produces species $i$ and negative when it consumes species $i$. The amount, stream flow, and generation rate can then be written as:

$$
\begin{aligned}
n_i &= c_iV,\\
\dot n_{i,s} &= c_{i,s}\dot V_s,\\
\dot n_{G,i} &= V\sum_{j\in\mathcal{R}}\sigma_{ij}v_j.
\end{aligned}
$$

Substituting these expressions into the mole balance and applying the product rule gives:

$$
\begin{aligned}
\frac{d}{dt}(c_iV)
&=\sum_{s\in\mathcal{F}}d_sc_{i,s}\dot V_s
  +V\sum_{j\in\mathcal{R}}\sigma_{ij}v_j,\\
V\frac{dc_i}{dt}+c_i\frac{dV}{dt}
&=\sum_{s\in\mathcal{F}}d_sc_{i,s}\dot V_s
  +V\sum_{j\in\mathcal{R}}\sigma_{ij}v_j.
\end{aligned}
$$

Every term in these amount balances has units $\mathrm{mmol\,h^{-1}}$. Dividing by $V$ and solving for the concentration derivative gives:

$$
\frac{dc_i}{dt}
=\frac{1}{V}\sum_{s\in\mathcal{F}}d_sc_{i,s}\dot V_s
 +\sum_{j\in\mathcal{R}}\sigma_{ij}v_j
 -\frac{c_i}{V}\frac{dV}{dt}.
$$

The final term accounts for a changing physical volume. For example, adding solvent without adding species $i$ lowers its concentration even if its amount remains unchanged. Constant volume removes this term.

## Intracellular balances on a biomass basis
Intracellular metabolite measurements and fluxes are commonly reported per gram of dry cell mass, denoted $\mathrm{gDW}$. This changes the normalization basis. Let $B>0$ denote the biomass concentration in $\mathrm{gDW\,L^{-1}}$ and $\bar V>0$ the culture volume in $\mathrm{L}$. The total biomass $\mathcal{B}$ is given by:

$$
\mathcal{B}=B\bar V.
$$

The quantity $\mathcal{B}$ has units $\mathrm{gDW}$. Let $C_i=n_i/\mathcal{B}$ denote the intracellular amount of species $i$ per unit biomass in $\mathrm{mmol\,gDW^{-1}}$. Likewise, let $\hat v_j$ denote a reaction or membrane-transport flux in $\mathrm{mmol\,gDW^{-1}\,h^{-1}}$. Multiplication by $\mathcal{B}$ converts these quantities to total amounts and rates.

Consider a batch or fed-batch culture in which no cells enter or leave and biomass changes only through growth. We represent molecular transport across cell membranes, including active and passive transport, in the reaction set $\mathcal{R}$. A culture feed changes the extracellular medium; it is not a bulk fluid stream passing directly through the intracellular compartment. With this choice of system boundary, the intracellular amount balance is given by:

$$
\frac{d}{dt}(C_i\mathcal{B})
=\mathcal{B}\sum_{j\in\mathcal{R}}\sigma_{ij}\hat v_j,
\qquad i\in\mathcal{M}.
$$

Both sides have units $\mathrm{mmol\,h^{-1}}$. Membrane transport enters through the flux sum, so there is no separate stream term.

Let's expand the accumulation term before imposing any steady-state or constant-volume assumption:

$$
\begin{aligned}
\frac{d}{dt}(C_iB\bar V)
&=B\bar V\frac{dC_i}{dt}
 +C_i\bar V\frac{dB}{dt}
 +C_iB\frac{d\bar V}{dt},\\
\sum_{j\in\mathcal{R}}\sigma_{ij}\hat v_j
&=\frac{dC_i}{dt}
 +\frac{C_i}{B}\frac{dB}{dt}
 +\frac{C_i}{\bar V}\frac{d\bar V}{dt}.
\end{aligned}
$$

The second line follows by dividing the amount balance by $\mathcal{B}=B\bar V$. Under the no-biomass-flow assumption, the specific growth rate $\mu$ is the fractional rate of increase of total biomass, with units $\mathrm{h^{-1}}$:

$$
\mu\equiv\frac{1}{\mathcal{B}}\frac{d\mathcal{B}}{dt}
=\frac{1}{B}\frac{dB}{dt}
 +\frac{1}{\bar V}\frac{d\bar V}{dt}.
$$

Define the concentration vector $\mathbf{x}=(C_1,\ldots,C_{|\mathcal{M}|})^\top$, the flux vector $\hat{\mathbf{v}}=(\hat v_1,\ldots,\hat v_{|\mathcal{R}|})^\top$, and the stoichiometric matrix $\mathbf{S}\in\mathbb{R}^{|\mathcal{M}|\times|\mathcal{R}|}$ with entries $\sigma_{ij}$. Combining the biomass derivatives gives the dynamic intracellular balance:

$$
\boxed{\mathbf{S}\hat{\mathbf{v}}
=\frac{d\mathbf{x}}{dt}+\mu\mathbf{x}}.
$$

Each term has units $\mathrm{mmol\,gDW^{-1}\,h^{-1}}$. The growth-dilution term $\mu\mathbf{x}$ accounts for the production needed to maintain metabolite amounts per unit biomass as biomass increases.

For a fixed-volume batch culture, $d\bar V/dt=0$ and $\mu=(1/B)dB/dt$. For fed-batch operation, the culture-volume derivative must remain when computing $\mu$ from $B$. Both operations lead to the same intracellular balance. If cells enter or leave, their carried metabolites and biomass must be included in the corresponding amount balances before normalizing. In particular, constant biomass concentration in a chemostat does not imply zero cellular growth.

## Simplified Species Constraints
The [flux balance analysis primer](docs/Orth-NatBiotechnol-28-2010.pdf) uses the constraint $\mathbf{S}\hat{\mathbf{v}}=\mathbf{0}$. To obtain this constraint from the dynamic balance, we need to distinguish constant intracellular composition from negligible growth dilution.

### Balanced growth and the conventional constraint
During balanced growth, intracellular amounts per unit biomass remain constant, so $d\mathbf{x}/dt=\mathbf{0}$. The total intracellular amounts can still increase because the total biomass increases. Substituting this condition into the dynamic balance gives:

$$
\mathbf{S}\hat{\mathbf{v}}=\mu\mathbf{x}.
$$

Thus, net production replenishes the metabolite pools as the cells grow.

Conventional flux balance analysis neglects the explicit growth dilution of metabolite pools when it is small relative to their rates of production and consumption. Let $q_i>0$ denote a characteristic production or consumption rate for species $i$, in $\mathrm{mmol\,gDW^{-1}\,h^{-1}}$. A dimensionally consistent comparison is given by:

$$
\frac{|\mu C_i|}{q_i}\ll1.
$$

Whether this holds depends on the metabolite pool and the growth conditions being modeled.

> __Conventional flux balance constraint:__
>
> If the modeled intracellular pools are approximately stationary per unit biomass and their explicit growth-dilution demands can be neglected, the dynamic balance is approximated by:
>
> $$
> \boxed{\mathbf{S}\hat{\mathbf{v}}=\mathbf{0}}.
> $$
>
> This is a model constraint: for each balanced metabolite, production and consumption, including modeled transport, cancel. It does not require zero growth, zero exchange, or a constant culture volume.

Growth still consumes precursors. Conventional reconstructions represent these demands with a biomass pseudo-reaction, whose coefficients specify the material required to form new biomass. When normalized per gram of new biomass, the biomass reaction has flux units $\mathrm{h^{-1}}$ and coefficients in $\mathrm{mmol\,gDW^{-1}}$; its contribution to each metabolite balance still has units $\mathrm{mmol\,gDW^{-1}\,h^{-1}}$. [Benyamini et al. (2010)](https://doi.org/10.1186/gb-2010-11-4-r43) examine how neglecting intermediate-metabolite dilution can affect predictions.

### Exchange reactions and the system boundary
Does every species in the culture have to remain at steady state? No. The balance constraint applies to the metabolite pools represented by the rows of $\mathbf{S}$. Substrate concentrations in the surrounding medium may fall while secreted products accumulate.

A transport reaction connects represented compartments, for example by moving a metabolite between extracellular and intracellular pools. An exchange reaction connects a represented boundary metabolite to surroundings that are not explicitly balanced. For a boundary metabolite $M_i$, one possible exchange convention is given by:

$$
\emptyset\rightleftharpoons M_i.
$$

Here, $\emptyset$ denotes the unmodeled surroundings. With positive flux defined from left to right, the exchange column has a single coefficient $+1$ in the row for $M_i$, positive flux supplies that pool, and negative flux removes material. A model written in the reverse orientation has the opposite flux signs. Flux bounds specify which directions and rates are allowed.

In reconstructions with an explicit extracellular compartment, exchange and membrane transport are separate reactions. If that compartment is omitted, a boundary reaction can represent their net effect. These choices let the stoichiometric equations describe an open system. [Bordbar et al. (2014)](https://doi.org/10.1038/nrg3643) illustrate how the resulting network and flux bounds constrain the feasible metabolic states.

### Cell-free systems use a physical-volume basis
A cell-free reaction mixture has no growing cell population, so the biomass basis used above is inappropriate. Return instead to the physical-volume balance, with concentrations $c_i$ in $\mathrm{mmol\,L^{-1}}$ and reaction rates $v_j$ in $\mathrm{mmol\,L^{-1}\,h^{-1}}$. For a constant-volume batch with no material exchange, define $\mathbf{c}=(c_i)_{i\in\mathcal{M}}$ and $\mathbf{v}=(v_j)_{j\in\mathcal{R}}$. Its balance reduces to:

$$
\mathbf{S}\mathbf{v}=\frac{d\mathbf{c}}{dt}.
$$

Substrates are consumed and products accumulate, so a zero-balance approximation requires a separate steady-state or quasi-steady-state assumption for the selected pools. Continuous cell-free systems must also retain material exchange, either as explicit flow terms or as equivalent boundary fluxes.

## Summary
We derived the conventional flux balance constraint from an amount balance by choosing a normalization basis, expanding the accumulation term, and stating which dynamic terms we neglect. This separates the intracellular approximation from reactor operating conditions and from the choice of system boundary.

> __Key Takeaways:__
>
> * **Two bases give two accumulation terms:** We derived the species balance per unit volume and per unit biomass from the same open mole balance. The volume basis picks up a changing-volume term, and the biomass basis picks up a growth-dilution term.
> * **Balanced growth alone does not zero the balance:** We showed that constant amounts per unit biomass still require net production to replenish the pools as the cells grow. The conventional constraint follows only when that growth dilution is small compared with each pool's production and consumption rates, while the biomass reaction keeps the growth demands.
> * **Exchange reactions make the system open:** We used transport and exchange reactions to carry material across the modeled boundary, so the intracellular balance holds without a steady extracellular environment. A cell-free system returns to the volume basis and needs its own accumulation and flow assumptions.

These results identify the assumptions behind the material-balance constraint used with flux bounds and an objective in the main lecture.
