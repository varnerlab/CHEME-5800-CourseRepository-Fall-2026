# L3d Algorithm: Bubblesort

Bubblesort is a simple method for ordering a list. It repeatedly passes over the list, compares adjacent items, and swaps any two neighbors that are out of order, until a pass finds nothing left to swap. It is worth studying not because you would use it, but because you can count its cost exactly.

Let's look at some pseudo-code for Bubblesort:

__Initialization:__ The algorithm starts with an unsorted list of elements in the vector $\mathbf{a}$. Compute the length of the unsorted list $n \gets \texttt{length}(a)$.

Repeat the following steps (step 4 decides when to stop):
1. Set a flag $\text{swapped} \gets \texttt{false}$.
2. For each $i\in\left\{1,\ldots,n-1\right\}$ __do__:
    * If $a_{i} > a_{i+1}$ __then__:
        * Call the $\texttt{swap}(a_{i}, a_{i+1})$ method to swap (exchange) the two elements.
        * Set the flag $\text{swapped} \gets \texttt{true}$ (indicating a swap was made).
3. Shrink the unsorted region, $n \gets n - 1$: after a pass, the largest remaining element has reached the end and never needs to be examined again.
4. If $\text{swapped}$ is $\texttt{false}$, the vector $\mathbf{a}$ is already ordered and the algorithm terminates. Otherwise, go back to step 1.

The $\texttt{swap}(a_{i}, a_{i+1})$ method does the following:

__Initialize__: Given two elements $a_{i}$ and $a_{i+1}$, we will swap them if they are out of order. The algorithm starts with the first element $a_{i}$ and the second element $a_{i+1}$.
1. Set a temporary variable $t \gets a_{i}$.
2. Set $a_{i} \gets a_{i+1}$.
3. Set $a_{i+1} \gets t$.
4. Return the swapped elements $a_{i}$ and $a_{i+1}$.

The test $a_{i} > a_{i+1}$ is written above in numeric terms. The implementation you complete in the lab asks [the `isless(...)` function](https://docs.julialang.org/en/v1/base/base/#Base.isless) instead, which is the ordering [the `sort(...)` function](https://docs.julialang.org/en/v1/base/sort/#Base.sort) uses, and the two agree for ordinary numbers.

## Worst-case time complexity
The worst-case time complexity of Bubblesort is $\mathcal{O}(n^2)$, where $n$ is the number of elements in the unsorted list. Let's see where that bound comes from.

### Analysis
The number of comparisons is $n-1$ for the first pass, $n-2$ for the second pass, and so on, until we reach 1 comparison. This gives us a total of $n-1 + n-2 + \ldots + 1 = \frac{(n-1)(n)}{2}$ comparisons. Further, the number of swaps is at most the number of comparisons, and in the worst case every comparison finds its pair out of order, so a swap follows each one. Thus, in that case we make $\frac{(n-1)(n)}{2}$ swaps.

Let $c$ be the time for one comparison together with the inner-loop bookkeeping that accompanies it, and let $s$ be the time for one swap together with the flag assignment it triggers (both constant). The only work these constants do not absorb is the setup at the top of each pass, which happens at most $n$ times.

Writing $T(n)$ for the comparison and swap time, we have:
$$\begin{align*}
T(n) &= \underbrace{c\cdot\left(\frac{(n-1)(n)}{2}\right)}_{\text{comparisons}} + \underbrace{s\cdot\left(\frac{(n-1)(n)}{2}\right)}_{\text{swaps}} \\
&= c\cdot\frac{n^2 - n}{2} + s\cdot\frac{n^2 - n}{2} \\
&= \left(c + s\right)\cdot\frac{n^2 - n}{2} \\
&\in \mathcal{O}(n^2)\,\blacksquare
\end{align*}$$

The per-pass setup left outside $c$ and $s$ adds a term linear in $n$, which the quadratic term dominates, so the bound is unchanged.

The best case is different, and the early exit in step 4 is what creates it: handed an already-ordered list, the first pass performs no swap, the algorithm stops, and the cost is the $n-1$ comparisons of that single pass.

## Summary
Bubblesort orders a list by repeatedly exchanging out-of-order neighbors, and you get both its quadratic worst case and its linear best case by counting the comparisons those passes make.

> __Key Takeaways:__
>
> * __Each pass finalizes one element:__ After a pass, the largest remaining value has reached the end of the unsorted region. That is why the region can shrink by one every time, and why one pass for each element after the first is enough to order the whole list.
> * __The swap flag turns the best case linear:__ A pass with no swaps proves the list is already ordered. The algorithm can therefore stop after a single pass on sorted input, instead of always running all of its passes.
> * __Simplicity is what it buys you:__ The algorithm is short enough to check by reading it, which is what makes it a useful teaching baseline. The early exit helps on input that is already ordered or nearly so, but the worst case stays quadratic, and that is what rules the algorithm out as the data grows.
