## Background: The shape of a recursive function
Recursion is a programming technique in which a function calls itself with a modified version of its input. Why use recursion? The key idea of recursion is to break a problem down into smaller subproblems. Each recursive call works on a smaller piece of the original problem until it reaches the base case.

Skeleton of a recursive function:
> 
> ```julia
> function recursive_function(input)
>    if base_case_condition(input) == true
>        return base_case_result
>    else
>        # Recursive case: call the function with a modified input
>        return recursive_function(modified_input)
>    end
> end
> ```

The _body_ of a recursive function has two parts:
* __Base case:__ This is the condition that stops the recursion. When the function reaches the base case, it returns a result without calling itself again. If a chain of calls does not move toward the base case, that chain never stops, and the run ends in a stack overflow instead of an answer.
* __Recursive case:__ This is the part of the function that calls itself with a modified version of its input. The recursive case breaks the problem into smaller subproblems and solves them with the same function.

Raising a number to an integer power is a classic example of recursion, and the figure shows the calls it makes: each level takes off one factor of $x$ and passes the smaller exponent to the next call.

Suppose we wanted to calculate $x^{n}$, where $n$ is a nonnegative integer. We can break this into a series of recursive calls. Mathematically, we can express this as:
$$
\begin{align*}
x^{n} &= x \cdot \underbrace{x^{n-1}}_{\text{recursion case}}\quad\text{if } n > 0 \\
x^{n} & =x\cdot\left(x\cdot\underbrace{x^{n-2}}_{\text{recursion case}}\right)\quad\text{if } n > 1 \\
x^{n} & = x\cdot\left(x\cdot\left(x\cdot\underbrace{x^{n-3}}_{\text{recursion case}}\right)\right)\quad\text{if } n>2 \\
x^{n} & = x\cdot\left(x\cdot\left(x\cdot\left(x\cdot\underbrace{x^{n-4}}_{\text{recursion case}}\right)\right)\right)\quad\text{if } n>3 \\
\dots & \\
\end{align*}
$$

### Base and Recursive cases
What are the two cases we need to consider?
* __Base case__: We need to have a base case that stops the recursion. In this case, we can say that if $n=0$, then we return $x^{0}=1$. This is our base case.
* __Recursive case__: If $n>0$, then we can say that $x^{n}=x\cdot x^{n-1}$. Thus, we can keep breaking the problem down with the $x^{n-1}$ term until we reach the base case.

Let's put this together in some pseudo-code:

__Initialization__: Given a number $x\in\mathbb{R}$ and an integer $n\in\mathbb{Z}_{\geq{0}}$, we want to compute $x^{n}$.

1. If $n=0$, then return 1 (base case).
2. If $n>0$, then return $x\cdot\text{power}(x,n-1)$ (recursive case).

That is recursion: the definition refers to itself, and the base case stops it. Let's implement this and see how it works in practice.

````julia
"""
    mypower(x, n, level = 0)

Computes x raised to the power of n using recursion, printing the level of each call.

### Arguments
- `x`: The base number to be raised to a power. No type is enforced here, and the base case returns `1.0`, so the result is whatever multiplying `x` into a `Float64` produces.
- `n`: The exponent to which the base number is raised. It must be a nonnegative integer, since a negative exponent moves away from the `n == 0` base case and recurses until the stack overflows.
- `level`: The current recursion level, which is used only for the printed trace. 

### Returns
- The value of x raised to the power of n, built up as repeated multiplication on the way back from the base case.
"""
function mypower(x, n, level = 0)

    ## Error checking would normally go here!
    ## ....

    if n == 0
        println("Base case reached: $x^$n = 1 at level $level")
        return 1.0 # base case: any number to the power of 0 is 1
    else

        println("Recursive case: $x^$n = $x * $x^$(n - 1) at level $level")
        smaller = mypower(x, n - 1, level + 1) # recursive case: solve the smaller problem first
        result = x * smaller
        println("Combining on the way back up: $x * $smaller = $result at level $level")
        return result
    end
end;
````

__Test__: Let's test our `mypower(...)` function. The [`println(...)` calls](https://docs.julialang.org/en/v1/base/io-network/#Base.println) inside the implementation report each level of the recursion as it runs, so the output traces the descent to the base case and then the multiplications that combine the returned values on the way back up. Since this is a test function we will not release to users, we skip the error checking.

For reference, here is the value we are looking for, computed with Julia's exponentiation operator:

````julia
2^10
````

Now the same value, computed by recursion instead and stored in the `value::Float64` variable. Watch the levels count up as the calls go down, the base case print at the bottom, and the combining lines print as each call returns:

````julia
value = let 

    # initialize -
    x = 2.0; # base number
    n = 10; # exponent

    # call the function
    value = mypower(x, n, 0) # should return x^n
end
````

## Understanding Recursion

Recursion is hard to follow at first. The part that confuses most readers is how the results of the individual calls are combined into the final result. A __tree__ is a good way to picture it. 

We cover trees in a few lectures. For now, take each recursive call to be a node in a tree. The tree shows how the calls are made and how the return values are combined.

> __What is a recursive call tree?__
> 
> Each recursive call is a node in a tree, and an edge connects a call to the smaller calls it makes. The figure shows the general shape for the __Fibonacci recurrence__, $F_{n} = F_{n-1} + F_{n-2}$ with base cases $F_{0}=0$ and $F_{1}=1$, where each call on $n$ makes two smaller calls. Calls go down the tree until every branch reaches a __base case__. 
> 
> The __base case__ is the condition that stops the recursion, and it returns a specific value. The returned values are then combined on the way back up. 

__Recursion is not always efficient.__ For example, follow the shaded nodes in the figure: the call on $n-2$ is made by the root and made again by the call on $n-1$, and the two calls on $n-3$ repeat yet another subproblem. Nothing in the plain recursive definition stops this recomputation.

However, what if we could save the results of each subproblem in a cache, so that when we encounter the same subproblem again, we can look it up instead of computing it again? This is called __memoization__, and it can make a recursive algorithm much faster.

The key is to break the problem into smaller subproblems and to define the base case and the recursive case clearly. Once you understand how it works, recursion becomes a useful tool, and it often states a problem in fewer lines than a loop does.

## Summary
A recursive function defines a result in terms of a smaller instance of the same problem, and a base case that stops it.

> __Key Takeaways:__
>
> * __A recursive function describes a result, it does not loop:__ It states an answer in terms of a smaller version of the same problem, plus a case that needs no further calls. That is why the definition stays short even when the work it produces is long.
> * __The base case is what stops it:__ Every call has to move closer to a case that returns without calling again. A base case that some inputs skip past is the usual reason a recursion never ends.
> * __Clarity and cost are separate questions:__ A recursive definition can be the clearest way to state a problem and still be the wrong way to compute it. Nothing in the definition stops the call tree from solving the same subproblem many times, which is what the companion example measures and then removes with a cache.

The `mypower(...)` trace and the Fibonacci call tree teach the same lesson at two sizes: a short definition turns into a computation you need a tree to follow, and once that tree starts repeating subproblems you should switch to a loop or add a cache.
